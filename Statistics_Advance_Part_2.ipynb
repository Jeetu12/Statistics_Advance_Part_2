# **Statistics Advance Part 2 : Theory Questions**

### 1. **What is hypothesis testing in statistics?**

- **Hypothesis testing** is a statistical method used to make decisions or inferences about a population based on sample data.

- It involves formulating two competing hypotheses:  

   - **Null Hypothesis (H_0)**: Assumes no effect or no difference in the population.

   - **Alternative Hypothesis (H_a)**: Represents the presence of an effect or difference.

- The goal of hypothesis testing is to determine whether there is enough statistical evidence in the sample data to reject (H_0) in favor of (H_a).

- **Steps in hypothesis testing:**

   1. Formulate (H_0) and (H_a).
   2. Choose a significance level (alpha).
   3. Select the appropriate test (e.g., Z-test, T-test).
   4. Compute the test statistic.
   5. Compare with the critical value or compute the p-value.
   6. Make a decision (reject or fail to reject (H_0)).

---

### 2. **What is the null hypothesis, and how does it differ from the alternative hypothesis?**

- The **null hypothesis (H_0)** assumes that there is no effect, no difference, or no relationship in the population. It is the default assumption.

- The **alternative hypothesis (H_a)** contradicts (H_0) and suggests that there is a significant effect or difference.

- **Example:**  

   - (H_0): The average height of male and female students is the same.  

   - (H_a): The average height of male and female students is different.

- If statistical evidence supports (H_a), we reject (H_0); otherwise, we fail to reject (H_0).

---

### 3. **What is the significance level in hypothesis testing, and why is it important?**

- The **significance level (alpha)** is the probability of rejecting the null hypothesis when it is actually true (Type I error). It represents the threshold for deciding whether a result is statistically significant.

- **Common significance levels:**

   - **0.05 (5%)**: There is a 5% chance of making a Type I error.
   
   - **0.01 (1%)**: Stricter level, used in critical research.

- **Importance:**
  
   - Determines the confidence in the decision.
  
   - Helps balance between Type I and Type II errors.

---

### 4. **What does a P-value represent in hypothesis testing?**
   
- The **P-value** is the probability of obtaining test results at least as extreme as the observed data, assuming (H_0) is true.

- **Interpretation:**

   - A **low P-value** (< significance level) suggests strong evidence against (H_0), leading to its rejection.
  
   - A **high P-value** (> significance level) means insufficient evidence to reject (H_0).

- **Example:**  

 - If (P = 0.03) and (alpha = 0.05), we reject (H_0) because
 (P < alpha).

---

### 5. **How do you interpret the P-value in hypothesis testing?**

- If **(P < alpha)** → Reject (H_0) (significant result).
   
- If **(P > alpha)** → Fail to reject (H_0) (not significant).

- **Example:**
   
  - (P = 0.02) and (alpha = 0.05) → Reject (H_0) (statistically significant).
  
  - (P = 0.07) and (alpha = 0.05) → Fail to reject (H_0) (not significant).

---

### 6. **What are Type 1 and Type 2 errors in hypothesis testing?**
- **Type I Error (alpha)**: Rejecting (H_0) when it is actually true (false positive).
   
- **Type II Error (beta)**: Failing to reject (H_0) when it is false (false negative).

- **Example:**
   
   - Type I: A healthy person is diagnosed as sick.
   
   - Type II: A sick person is diagnosed as healthy.

- **Trade-off**: Reducing one type of error increases the other.

---

### 7. **What is the difference between a one-tailed and a two-tailed test in hypothesis testing?**

- **One-tailed test**: Tests for an effect in one specific direction.
   
- **Two-tailed test**: Tests for an effect in both directions.

- **Example:**
  
   - One-tailed: "New drug increases lifespan" (H_a: mu > mu_0).
  
   - Two-tailed: "New drug changes lifespan" (H_a: mu != mu_0).

---

### 8. **What is the Z-test, and when is it used in hypothesis testing?**
   
- A **Z-test** is used to determine if there is a significant difference between sample and population means when:
   
   - The population variance is known.
   
   - The sample size is large (n > 30).

- **Example:**  
   
   - Checking if the average test score of students differs from the national average.

---

### 9. **How do you calculate the Z-score, and what does it represent in hypothesis testing?**
   
- **Z-score formula**:
   [
   Z = {X - mu}{sigma}
   ]
   Where:
   - (X) = Sample mean
   - (mu) = Population mean
   - (sigma) = Standard deviation

- The Z-score measures how many standard deviations a data point is from the mean.

---

### 10. **What is the T-distribution, and when should it be used instead of the normal distribution?**

- The **T-distribution** is used when:
  
   - The sample size is small (n < 30).
  
   - The population standard deviation (sigma) is unknown.

- It has heavier tails than the normal distribution, making it suitable for small sample sizes.

---

### 11. **What is the difference between a Z-test and a T-test?**

- **Z-test**: Used for large samples (n > 30) with known (sigma).
   
- **T-test**: Used for small samples (n < 30) with unknown (sigma).

---

### 12. **What is the T-test, and how is it used in hypothesis testing?**

- The **T-test** compares means between groups when population variance is unknown.

- **Types:**
  
   - **One-sample T-test**: Compares sample mean to population mean.
  
   - **Independent T-test**: Compares means of two independent groups.
  
   - **Paired T-test**: Compares means of the same group before and after treatment.

---

### 13. **What is the relationship between Z-test and T-test in hypothesis testing?**
   
- Both test hypotheses about population means.

- T-test is used for small samples, Z-test for large samples.
   
- As (n) increases, the T-distribution approaches the normal Z-distribution.

---

### 14. **What is a confidence interval, and how is it used to interpret statistical results?**
   
- A **Confidence Interval (CI)** is a range of values that is likely to contain the **true population parameter** (such as mean or proportion) with a certain level of confidence. It provides an **estimate of uncertainty** and is widely used in **statistical analysis**.

- If the **population variance is unknown**, we use the **T-distribution** instead of the Z-score.

1. **Estimation of Population Parameters**  

   - A confidence interval provides a **range** within which the **true population parameter** is likely to fall.  

2. **Decision Making in Hypothesis Testing**  

   - If the **null hypothesis value** (e.g., (mu_0) **falls outside the CI**, we reject (H_0)).  

3. **Comparing Groups**  

   - If two confidence intervals **do not overlap**, it suggests a **statistically significant difference** between groups.  

- ### **Example: Confidence Interval Calculation in Python**

```python
import numpy as np
import scipy.stats as stats

# Step 1: Generate Sample Data

np.random.seed(42)
sample = np.random.normal(loc=50, scale=10, size=30)  # Mean=50, Std=10, Sample Size=30

# Step 2: Compute Sample Statistics

sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)  # Sample standard deviation
n = len(sample)

# Step 3: Define Confidence Level

confidence_level = 0.95
alpha = 1 - confidence_level
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)  # T-score for 95% CI

# Step 4: Compute Margin of Error

margin_of_error = t_critical * (sample_std / np.sqrt(n))

# Step 5: Compute Confidence Interval

lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error

# Step 6: Print Results

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Confidence Interval (95%): ({lower_bound:.2f}, {upper_bound:.2f})")
```
- ### **Interpreting the Results:**

 - If the output is:  
  ```
  Sample Mean: 48.25
  Confidence Interval (95%): (44.87, 51.63)
  ```
  This means we are **95% confident** that the **true population mean** lies between **44.87 and 51.63**.

- ### **Key Takeaways:**

✅ **Confidence intervals measure uncertainty in estimating population parameters.**  

✅ **A 95% CI means that in repeated samples, 95% of the intervals will contain the true mean.**  

✅ **If the CI does not contain a hypothesized value (e.g., population mean), we reject the null hypothesis.**

### 15. **What is the margin of error, and how does it affect the confidence interval?**  

- The **margin of error (ME)** represents the range of uncertainty around a sample statistic, often in survey results or hypothesis testing. It is a measure of how much the sample result could differ from the true population value.  


- **Effect on confidence interval (CI):**  
  
 - A **smaller margin of error** → More precise estimate (narrower CI).  

 - A **larger margin of error** → Less precise estimate (wider CI).  

 - Increasing **sample size** reduces the margin of error, increasing accuracy.  

- **Example:**  

 - If a poll shows a candidate has 55% support with a margin of error of ±3%, the true support is likely between **52% and 58%**.

### 16. **How is Bayes' Theorem used in statistics, and what is its significance?**  

- **Bayes’ Theorem** describes how to update probabilities based on new evidence. It is fundamental in **probabilistic reasoning, machine learning, and medical diagnostics**.  


- **Example:**  

 - A medical test for a disease is 95% accurate, and the disease affects 1% of the population. Bayes’ Theorem helps calculate the probability that someone actually has the disease given a positive test result.  

- **Significance:**  

 - Used in spam filtering, fraud detection, and AI.  

 - Helps in decision-making under uncertainty.

### 17. **What is the Chi-square distribution, and when is it used?**  

- The **Chi-square (chi^2) distribution** is used in statistical tests dealing with categorical data. It measures how observed values differ from expected values.  

**Properties:**  

- Skewed right (especially for small degrees of freedom).  

- As degrees of freedom increase, it approaches a normal distribution.  

**Used in:**  

- **Chi-square goodness-of-fit test**: Tests if sample data fits an expected distribution.  

- **Chi-square test for independence**: Determines if two categorical variables are related.  

**Example:**  

- Testing if dice rolls follow a fair uniform distribution.  

---

### 18. **What is the Chi-square goodness-of-fit test, and how is it applied?**  

- The **goodness-of-fit test** checks if an observed frequency distribution matches an expected distribution.  

**Steps:**  

1. Define (H_0) (data follows expected distribution).  

2. Calculate expected frequencies.  

3. Compute (chi^2) statistic.  

4. Compare with critical value or P-value.  

5. If (chi^2) is large, reject (H_0).  

**Example:**  

- Checking if the number of students choosing different college majors matches historical data.  

---

### 19. **What is the F-distribution, and when is it used in hypothesis testing?**  

- The **F-distribution** is a ratio of two variances and is used in comparing variability between groups.  

**Properties:**  

- Always positive.  

- Skewed right.  

- Defined by two degrees of freedom (df_1, df_2).  

**Uses:**  

- **ANOVA (Analysis of Variance)**: Compares means of multiple groups.  

- **F-test**: Compares two population variances.  

**Example:**  

- Testing if the variance in test scores differs between schools.  

---

### 20. **What is an ANOVA test, and what are its assumptions?**  

- **ANOVA (Analysis of Variance)** tests if three or more group means are significantly different.  

- **Assumptions:**  

1. **Independence**: Observations must be independent.

2. **Normality**: Data should be normally distributed.  

3. **Equal variance (homoscedasticity)**: Variances should be roughly equal across groups.  

**Example:**  

- Comparing exam scores across three schools.  

---

### 21. **What are the different types of ANOVA tests?**  

1. **One-way ANOVA**: Compares means across one categorical variable.  

   - Example: Examining salary differences across three industries.  

2. **Two-way ANOVA**: Examines the effect of two categorical variables.  

   - Example: Studying the effect of gender and education level on salaries.  

3. **Repeated-measures ANOVA**: Used when the same subjects are measured multiple times.  

   - Example: Measuring blood pressure before, during, and after treatment.  

---

### 22. **What is the F-test, and how does it relate to hypothesis testing?**

- An **F-test** compares variances to determine if they are significantly different.  

- **Uses:**  

 - Testing if two datasets have the same variance.  

 - ANOVA (to compare multiple means).  

- **Example:**  

 - Comparing whether the variability in test scores differs between two schools.

````python
# 13. Generate random variables and calculate their corresponding
# probabilities using the binomial distribution.

import random
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

def binomial_probabilities(n, p):
    """
    Generates random variables and calculates their probabilities using the binomial distribution.

    Args:
      n: The number of trials.
      p: The probability of success on each trial.

    Returns:
      A tuple containing:
        - A list of random variables.
        - A list of corresponding probabilities.
    """
    random_variables = []
    probabilities = []

    for _ in range(n):
        k = random.randint(0, n) # Generate random variable between 0 and n.
        probability = binom.pmf(k, n, p) # calculate probability using binom.pmf().
        random_variables.append(k)
        probabilities.append(probability)

    return random_variables, probabilities

# Example usage
n = 10  # Number of trials
p = 0.5 # Probability of success


random_vars, probs = binomial_probabilities(n, p)
for i in range(n):
    print(f"Random Variable: {random_vars[i]}, Probability: {probs[i]}")


# Plot histogram
plt.hist(random_vars, bins=np.arange(0, n+2)-0.5, density=True,
         color='skyblue', edgecolor='black')
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title(f"Binomial Distribution (n={n}, p={p})")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
````

````python
# 14. Write a Python program to calculate the Z-score for a given data point and
# compare it to a standard normal distribution.

import numpy as np
from scipy.stats import norm

def calculate_zscore(data_point, data_mean, data_std):
    """Calculates the Z-score for a given data point.

    Args:
        data_point: The data point for which to calculate the Z-score.
        data_mean: The mean of the dataset.
        data_std: The standard deviation of the dataset.

    Returns:
        The Z-score of the data point.
    """
    z_score = (data_point - data_mean) / data_std
    return z_score

def compare_to_standard_normal(z_score):
  """Compares the calculated z-score to standard normal distrubution

  Args:
      z_score: the z-score to be compared to standard normal distribution

  Returns:
      probability: cumulative probability of z-score in standard normal distribution
  """
  probability = norm.cdf(z_score)
  return probability

# Example Usage
data_point = 75
data_mean = 70
data_std = 5

# Calculate the Z-score
z_score = calculate_zscore(data_point, data_mean, data_std)
print(f"The Z-score of the data point {data_point} is: {z_score}")

# compare the z-score with the standard normal distribution
probability = compare_to_standard_normal(z_score)
print(f"The probability of a z-score of {z_score} or less in the standard normal distribution is: {probability}")
````

````python
# 15. Implement hypothesis testing using Z-statistics for a sample dataset.

import numpy as np
import scipy.stats as stats

def z_test(sample, population_mean, population_std, alpha=0.05, alternative="two-sided"):
    """
    Perform a Z-test for a sample dataset.

    Parameters:
    sample (list or numpy array): Sample data
    population_mean (float): Mean of the population
    population_std (float): Standard deviation of the population
    alpha (float): Significance level (default: 0.05)
    alternative (str): Type of test - "two-sided", "greater", or "less"

    Returns:
    dict: Z-score, p-value, and conclusion
    """
    sample_mean = np.mean(sample)
    sample_size = len(sample)
    standard_error = population_std / np.sqrt(sample_size)

    # Compute Z-score
    z_score = (sample_mean - population_mean) / standard_error

    # Compute p-value based on the alternative hypothesis
    if alternative == "two-sided":
        p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
    elif alternative == "greater":
        p_value = 1 - stats.norm.cdf(z_score)
    elif alternative == "less":
        p_value = stats.norm.cdf(z_score)
    else:
        raise ValueError("Invalid alternative hypothesis. Choose 'two-sided', 'greater', or 'less'.")

    # Conclusion
    reject_null = p_value < alpha
    conclusion = "Reject the null hypothesis" if reject_null else "Fail to reject the null hypothesis"

    return {
        "Z-score": z_score,
        "p-value": p_value,
        "Conclusion": conclusion
    }

# Example usage
sample_data = [52, 48, 50, 47, 53, 49, 51, 50, 54, 46]
pop_mean = 50
pop_std = 3
alpha = 0.05

result = z_test(sample_data, pop_mean, pop_std, alpha, alternative="two-sided")
print(result)
````

````python
# 16. Create a confidence interval for a dataset using Python and
# interpret the result.

import numpy as np
import scipy.stats as stats

def confidence_interval(sample, confidence=0.95, population_std=None):
    """
    Compute the confidence interval for a dataset.

    Parameters:
    sample (list or numpy array): Sample data
    confidence (float): Confidence level (default: 0.95)
    population_std (float, optional): Known population standard deviation. If None, use sample standard deviation.

    Returns:
    tuple: (Lower bound, Upper bound) of the confidence interval
    """
    sample_mean = np.mean(sample)
    sample_size = len(sample)

    # If population standard deviation is known, use it; otherwise, use sample standard deviation
    if population_std is None:
        sample_std = np.std(sample, ddof=1)  # Sample standard deviation (unbiased)
        standard_error = sample_std / np.sqrt(sample_size)
        t_critical = stats.t.ppf((1 + confidence) / 2, df=sample_size - 1)  # t-score for t-distribution
        margin_of_error = t_critical * standard_error
    else:
        standard_error = population_std / np.sqrt(sample_size)
        z_critical = stats.norm.ppf((1 + confidence) / 2)  # Z-score for normal distribution
        margin_of_error = z_critical * standard_error

    # Confidence Interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound

# Example usage
sample_data = [52, 48, 50, 47, 53, 49, 51, 50, 54, 46]
confidence_level = 0.95
population_std = 3  # Known population standard deviation (optional)

ci = confidence_interval(sample_data, confidence=confidence_level, population_std=population_std)
print(f"Confidence Interval: {ci}")
````

````python
# 2. Simulate random data to perform hypothesis testing and calculate the
# corresponding P-value using Python.

import numpy as np
import scipy.stats as stats

# Function to perform hypothesis testing
def hypothesis_test(sample1, sample2, alpha=0.05):
    # Calculate sample statistics
    mean1, mean2 = np.mean(sample1), np.mean(sample2)
    std1, std2 = np.std(sample1, ddof=1), np.std(sample2, ddof=1)
    n1, n2 = len(sample1), len(sample2)

    # Choose Z-test or T-test
    if n1 >= 30 and n2 >= 30:
        # Perform two-sample Z-test (Assume known population std)
        se = np.sqrt((std1**2 / n1) + (std2**2 / n2))
        z_score = (mean1 - mean2) / se
        p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
        test_type = "Z-Test"
    else:
        # Perform two-sample T-test (for small samples)
        t_stat, p_value = stats.ttest_ind(sample1, sample2, equal_var=False)
        test_type = "T-Test"

    return test_type, mean1, mean2, p_value

# Step 1: Simulate random data from two normal distributions
np.random.seed(42)
sample1 = np.random.normal(loc=100, scale=15, size=50)  # Mean=100, Std=15, Size=50
sample2 = np.random.normal(loc=105, scale=15, size=50)  # Mean=105, Std=15, Size=50

# Step 2: Perform hypothesis testing
test_type, mean1, mean2, p_value = hypothesis_test(sample1, sample2)

# Step 3: Interpret Results
alpha = 0.05  # Significance level

print(f"Test Used: {test_type}")
print(f"Sample 1 Mean: {mean1:.2f}")
print(f"Sample 2 Mean: {mean2:.2f}")
print(f"P-Value: {p_value:.4f}")

if p_value < alpha:
    print("Result: Reject the null hypothesis (Significant difference between samples)")
else:
    print("Result: Fail to reject the null hypothesis (No significant difference between samples)")
````

````python
# 5. Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing.

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def calculate_and_visualize_errors(true_positive_rate, false_positive_rate, sample_size=1000):
    """
    Calculates and visualizes Type 1 and Type 2 errors in hypothesis testing.

    Args:
        true_positive_rate: The probability of correctly rejecting the null hypothesis
                            when it is false (1 - β, power of the test).
        false_positive_rate: The probability of incorrectly rejecting the null hypothesis
                             when it is true (α, significance level).
        sample_size: The size of the sample used in the simulation.
    """

    # Simulate data under null hypothesis (H0: true mean = 0)
    null_data = np.random.normal(0, 1, sample_size)

    # Simulate data under alternative hypothesis (H1: true mean ≠ 0)
    # We assume a small effect size for illustration
    alternative_data = np.random.normal(0.5, 1, sample_size)

    # Calculate critical values based on false_positive_rate (α)
    critical_value = norm.ppf(1 - false_positive_rate / 2)  # For a two-tailed test

    # Calculate Type 1 and Type 2 error rates
    type1_error = np.mean(np.abs(null_data) > critical_value)  # Observe errors in null data
    type2_error = np.mean(np.abs(alternative_data) <= critical_value) # Observe errors in alternative data


    print(f"Type 1 Error Rate (α): {type1_error:.4f}")
    print(f"Type 2 Error Rate (β): {type2_error:.4f}")
    print(f"Power (1-β): {1 - type2_error:.4f}")

    # Visualization
    plt.figure(figsize=(10, 6))

    # Plot distributions
    x = np.linspace(-4, 4, 500)
    plt.plot(x, norm.pdf(x, 0, 1), label='H0: Null Distribution')
    plt.plot(x, norm.pdf(x, 0.5, 1), label='H1: Alternative Distribution')  # Alternative mean is slightly shifted

    # Shade rejection regions
    plt.fill_between(x, 0, norm.pdf(x, 0, 1), where=x >= critical_value, color='red', alpha=0.3, label='Rejection Region')  # Right tail
    plt.fill_between(x, 0, norm.pdf(x, 0, 1), where=x <= -critical_value, color='red', alpha=0.3)  # Left tail

    # Mark the critical values
    plt.axvline(critical_value, color='black', linestyle='--', label=f"Critical Value = {critical_value:.2f}")
    plt.axvline(-critical_value, color='black', linestyle='--')


    plt.xlabel("Test Statistic")
    plt.ylabel("Probability Density")
    plt.title("Type 1 and Type 2 Errors Visualization")
    plt.legend()
    plt.grid(True)
    plt.show()

# Example Usage:
calculate_and_visualize_errors(true_positive_rate=0.8, false_positive_rate=0.05)
````

````python
# 6. Write a Python program to perform an independent T-test and interpret the results.

from scipy.stats import ttest_ind

def perform_independent_t_test(data1, data2, alpha=0.05):
    """
    Performs an independent two-sample t-test and interprets the results.

    Args:
        data1: The first dataset (list or numpy array).
        data2: The second dataset (list or numpy array).
        alpha: Significance level (default is 0.05).

    Returns:
        A dictionary containing the t-statistic, p-value, and interpretation.
    """

    t_statistic, p_value = ttest_ind(data1, data2)

    if p_value < alpha:
        interpretation = "Reject the null hypothesis: There is a statistically significant difference between the means of the two groups."
    else:
        interpretation = "Fail to reject the null hypothesis: There is no statistically significant difference between the means of the two groups."

    return {
        "t-statistic": t_statistic,
        "p-value": p_value,
        "interpretation": interpretation
    }


# Example Usage
data_group1 = [10, 12, 15, 11, 13]
data_group2 = [8, 9, 10, 11, 9]

results = perform_independent_t_test(data_group1, data_group2)
results
````

````python
# 8. Simulate data and perform both Z-test and T-test, then compare the results using Python.

import numpy as np
import scipy.stats as stats

# Function to perform a Z-test
def z_test(sample1, sample2):
    mean1, mean2 = np.mean(sample1), np.mean(sample2)
    std1, std2 = np.std(sample1, ddof=1), np.std(sample2, ddof=1)
    n1, n2 = len(sample1), len(sample2)

    # Compute standard error
    se = np.sqrt((std1**2 / n1) + (std2**2 / n2))

    # Compute Z-score
    z_score = (mean1 - mean2) / se

    # Compute p-value (two-tailed)
    p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

    return z_score, p_value

# Function to perform a T-test
def t_test(sample1, sample2):
    t_stat, p_value = stats.ttest_ind(sample1, sample2, equal_var=False)  # Welch's T-test (unequal variance)
    return t_stat, p_value

# Step 1: Simulate two datasets
np.random.seed(42)
sample_size_small = 25  # Small sample (T-test more relevant)
sample_size_large = 100  # Large sample (Z-test valid)
pop_mean1, pop_std1 = 100, 15
pop_mean2, pop_std2 = 105, 15

# Generate two samples
small_sample1 = np.random.normal(pop_mean1, pop_std1, sample_size_small)
small_sample2 = np.random.normal(pop_mean2, pop_std2, sample_size_small)

large_sample1 = np.random.normal(pop_mean1, pop_std1, sample_size_large)
large_sample2 = np.random.normal(pop_mean2, pop_std2, sample_size_large)

# Step 2: Perform Z-test and T-test on small and large samples
z_score_small, p_value_z_small = z_test(small_sample1, small_sample2)
t_stat_small, p_value_t_small = t_test(small_sample1, small_sample2)

z_score_large, p_value_z_large = z_test(large_sample1, large_sample2)
t_stat_large, p_value_t_large = t_test(large_sample1, large_sample2)

# Step 3: Print results
print("### Small Sample (n=25 per group) ###")
print(f"Z-Test: Z-Score = {z_score_small:.2f}, P-Value = {p_value_z_small:.4f}")
print(f"T-Test: T-Statistic = {t_stat_small:.2f}, P-Value = {p_value_t_small:.4f}")
print("-------------------------------------------------")

print("### Large Sample (n=100 per group) ###")
print(f"Z-Test: Z-Score = {z_score_large:.2f}, P-Value = {p_value_z_large:.4f}")
print(f"T-Test: T-Statistic = {t_stat_large:.2f}, P-Value = {p_value_t_large:.4f}")
````

````python
# 9. Write a Python function to calculate the confidence interval for a sample mean and explain its significance.

import numpy as np
import scipy.stats as stats

# Function to compute confidence interval for a sample mean
def confidence_interval(sample, confidence=0.95):
    sample_mean = np.mean(sample)
    sample_std = np.std(sample, ddof=1)  # Sample standard deviation
    n = len(sample)

    # Compute the margin of error (using T-distribution for small samples)
    t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)
    margin_of_error = t_critical * (sample_std / np.sqrt(n))

    # Confidence Interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return sample_mean, (lower_bound, upper_bound)

# Step 1: Generate a sample dataset
np.random.seed(42)
sample_size = 30
pop_mean = 100
pop_std = 15
sample = np.random.normal(loc=pop_mean, scale=pop_std, size=sample_size)

# Step 2: Compute 95% Confidence Interval
sample_mean, ci = confidence_interval(sample)

# Step 3: Print Results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"95% Confidence Interval: ({ci[0]:.2f}, {ci[1]:.2f})")
````

````python
# 11. Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process.

def bayes_theorem(prior_A, likelihood_B_given_A, prior_not_A, likelihood_B_given_not_A):
    """
    Computes the posterior probability using Bayes' Theorem.

    Parameters:
    prior_A : float  -> Prior probability P(A)
    likelihood_B_given_A : float  -> Likelihood P(B|A)
    prior_not_A : float  -> Complement of prior probability P(~A)
    likelihood_B_given_not_A : float  -> Likelihood P(B|~A)

    Returns:
    posterior_A_given_B : float  -> Posterior probability P(A|B)
    """

    # Compute total probability of evidence P(B)
    P_B = (likelihood_B_given_A * prior_A) + (likelihood_B_given_not_A * prior_not_A)

    # Compute posterior probability P(A|B)
    posterior_A_given_B = (likelihood_B_given_A * prior_A) / P_B

    return posterior_A_given_B

# Example: Medical Diagnosis (Bayesian Inference)
# ----------------------------------------------
# Suppose a rare disease affects 1% of a population.
# A test correctly identifies the disease 95% of the time (True Positive Rate).
# However, the test gives a false positive 5% of the time.

# Given data:
prior_disease = 0.01  # P(Disease)
likelihood_positive_given_disease = 0.95  # P(Test+ | Disease)
prior_no_disease = 1 - prior_disease  # P(No Disease)
likelihood_positive_given_no_disease = 0.05  # P(Test+ | No Disease)

# Compute the probability of having the disease given a positive test result
posterior_disease_given_positive = bayes_theorem(
    prior_disease,
    likelihood_positive_given_disease,
    prior_no_disease,
    likelihood_positive_given_no_disease
)

# Print the result
print(f"Probability of having the disease given a positive test result: {posterior_disease_given_positive:.4f}")
````

````python
# 16. Implement an F-test using Python to compare the variances of two random samples.

import numpy as np
import scipy.stats as stats

# Step 1: Generate two random samples
np.random.seed(42)  # For reproducibility
sample1 = np.random.normal(loc=50, scale=10, size=30)  # Mean=50, Std=10
sample2 = np.random.normal(loc=50, scale=15, size=30)  # Mean=50, Std=15

# Step 2: Calculate sample variances
var1 = np.var(sample1, ddof=1)  # Sample variance (unbiased)
var2 = np.var(sample2, ddof=1)  # Sample variance (unbiased)

# Step 3: Compute F-statistic (F-test)
F_statistic = var1 / var2  # Ensure F > 1
df1 = len(sample1) - 1
df2 = len(sample2) - 1

# Step 4: Compute p-value
p_value = 2 * min(stats.f.cdf(F_statistic, df1, df2), 1 - stats.f.cdf(F_statistic, df1, df2))

# Step 5: Print results
print("F-Test for Equality of Variances")
print(f"Sample 1 Variance: {var1:.4f}")
print(f"Sample 2 Variance: {var2:.4f}")
print(f"F-Statistic: {F_statistic:.4f}")
print(f"Degrees of Freedom: ({df1}, {df2})")
print(f"P-Value: {p_value:.4f}")

# Step 6: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. The variances are significantly different.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference in variances.")
````

````python
# 19. Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA

def check_anova_assumptions(data, groups):
    """
    Checks the assumptions for ANOVA: normality, independence, and equal variance.

    Args:
      data: A list or numpy array of data values.
      groups: A list or numpy array of group labels corresponding to the data.

    Returns:
      A dictionary containing the results of the assumption checks.
    """

    import pandas as pd
    from scipy import stats
    import matplotlib.pyplot as plt

    # Create a DataFrame for easier analysis
    df = pd.DataFrame({'data': data, 'group': groups})

    # 1. Normality (Shapiro-Wilk test)
    normality_results = {}
    for group in df['group'].unique():
        group_data = df[df['group'] == group]['data']
        shapiro_statistic, shapiro_pvalue = stats.shapiro(group_data)
        normality_results[group] = {
            'Shapiro-Wilk Statistic': shapiro_statistic,
            'p-value': shapiro_pvalue
        }
        if shapiro_pvalue < 0.05:
            print(f"Warning: Group {group} may not be normally distributed.")

    # Visualization: Q-Q plot
    plt.figure(figsize=(8, 6))
    for group in df['group'].unique():
      stats.probplot(df[df['group'] == group]['data'], dist="norm", plot=plt)
    plt.title("Q-Q Plot for each group")
    plt.show()

    # 2. Independence:  Assumed in the input data structure.

    # 3. Homogeneity of Variance (Levene's Test)
    levene_statistic, levene_pvalue = stats.levene(*[df[df['group'] == group]['data'] for group in df['group'].unique()])

    if levene_pvalue < 0.05:
        print(f"Warning: Levene's Test indicates unequal variances.")

    return {
      'normality': normality_results,
      "homogeneity_of_variance": {
          'Levene Statistic': levene_statistic,
          'p-value': levene_pvalue,
      }
  }

# Example Usage:
data = [10, 12, 15, 11, 13, 8, 9, 10, 11, 9, 14, 16, 17, 15, 18]
groups = ['A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'B', 'C', 'C', 'C', 'C', 'C']
assumption_checks = check_anova_assumptions(data, groups)
assumption_checks
````

````python
# 20. Perform a two-way ANOVA test using Python to study the interaction between two factors and visualize the
# results.

import numpy as np
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Generate Sample Data
np.random.seed(42)

# Factor A (e.g., Teaching Method: Traditional, Online)
method = np.repeat(["Traditional", "Online"], 30)

# Factor B (e.g., Test Type: MCQ, Descriptive)
test_type = np.tile(["MCQ", "Descriptive"], 30)

# Generate Scores (Dependent Variable) with interaction effect
scores = np.random.normal(loc=70, scale=10, size=60)  # Base scores
scores[method == "Online"] += 5  # Online method increases scores
scores[test_type == "MCQ"] += 3  # MCQ test type increases scores
scores[(method == "Online") & (test_type == "MCQ")] += 2  # Interaction effect

# Step 2: Create DataFrame
df = pd.DataFrame({"Method": method, "Test_Type": test_type, "Score": scores})

# Step 3: Perform Two-Way ANOVA
model = ols("Score ~ C(Method) + C(Test_Type) + C(Method):C(Test_Type)", data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)  # Type 2 ANOVA Table

# Step 4: Print ANOVA Results
print("Two-Way ANOVA Test Results:")
print(anova_table)

# Step 5: Visualization - Boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(x="Method", y="Score", hue="Test_Type", data=df, palette="Set2")
plt.title("Two-Way ANOVA: Effect of Method & Test Type on Scores")
plt.xlabel("Teaching Method")
plt.ylabel("Scores")
plt.grid(True)
plt.legend(title="Test Type")
plt.show()
````

````python
# 22. Perform a one-way ANOVA test in Python and visualize the results with boxplots to compare group means.

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Step 1: Generate Sample Data for Three Groups
np.random.seed(42)

group1 = np.random.normal(loc=50, scale=10, size=30)  # Mean=50, Std=10
group2 = np.random.normal(loc=55, scale=10, size=30)  # Mean=55, Std=10
group3 = np.random.normal(loc=60, scale=10, size=30)  # Mean=60, Std=10

# Combine data into a DataFrame for visualization
df = pd.DataFrame({
    "Score": np.concatenate([group1, group2, group3]),
    "Group": ["Group 1"] * len(group1) + ["Group 2"] * len(group2) + ["Group 3"] * len(group3)
})

# Step 2: Perform One-Way ANOVA Test
F_statistic, p_value = stats.f_oneway(group1, group2, group3)

# Step 3: Print ANOVA Results
print("One-Way ANOVA Test Results")
print(f"F-Statistic: {F_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")

# Step 4: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. There is a significant difference between group means.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference between group means.")

# Step 5: Visualization - Boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(x="Group", y="Score", data=df, palette="Set2")
plt.title("Comparison of Group Means using Boxplot")
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.grid(True)
plt.show()
````

````python
# 23. Simulate random data from a normal distribution, then perform hypothesis testing to evaluate the means.

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Simulate Two Independent Normal Distributions
np.random.seed(42)

group1 = np.random.normal(loc=50, scale=10, size=30)  # Mean=50, Std=10
group2 = np.random.normal(loc=55, scale=10, size=30)  # Mean=55, Std=10

# Step 2: Perform Independent Two-Sample t-Test
t_stat, p_value = stats.ttest_ind(group1, group2, equal_var=True)

# Step 3: Print Test Results
print("Two-Sample t-Test Results")
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

# Step 4: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. There is a significant difference between the group means.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference between the group means.")

# Step 5: Visualization - Boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(data=[group1, group2], palette="Set2")
plt.xticks([0, 1], ["Group 1", "Group 2"])
plt.title("Comparison of Group Means using Boxplot")
plt.ylabel("Values")
plt.grid(True)
plt.show()
````

````python
# 24.  Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results.

import numpy as np
import scipy.stats as stats

# Step 1: Generate Sample Data from Normal Distribution
np.random.seed(42)
sample = np.random.normal(loc=50, scale=10, size=30)  # Mean=50, Std=10

# Step 2: Define Hypothesized Population Variance
pop_variance = 100  # Assumed population variance (std^2)

# Step 3: Compute Sample Variance
sample_variance = np.var(sample, ddof=1)  # Sample variance with Bessel's correction

# Step 4: Compute Chi-Square Test Statistic
n = len(sample)
chi_square_stat = (n - 1) * sample_variance / pop_variance

# Step 5: Compute P-Value (Two-Tailed Test)
p_value = 2 * min(stats.chi2.cdf(chi_square_stat, df=n-1), 1 - stats.chi2.cdf(chi_square_stat, df=n-1))

# Step 6: Print Results
print("Chi-Square Test for Population Variance")
print(f"Sample Variance: {sample_variance:.4f}")
print(f"Chi-Square Statistic: {chi_square_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

# Step 7: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. The population variance is significantly different from the hypothesized value.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference in population variance.")
````

````python
# 25. Write a Python script to perform a Z-test for comparing proportions between two datasets or groups.

import numpy as np
import scipy.stats as stats

# Step 1: Define Sample Data
n1, x1 = 200, 50  # Group 1: Total = 200, Successes = 50
n2, x2 = 250, 80  # Group 2: Total = 250, Successes = 80

# Step 2: Compute Sample Proportions
p1 = x1 / n1  # Proportion in Group 1
p2 = x2 / n2  # Proportion in Group 2

# Step 3: Compute Pooled Proportion
p_pooled = (x1 + x2) / (n1 + n2)

# Step 4: Compute Standard Error
se = np.sqrt(p_pooled * (1 - p_pooled) * ((1/n1) + (1/n2)))

# Step 5: Compute Z-Score
z_score = (p1 - p2) / se

# Step 6: Compute P-Value (Two-Tailed Test)
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

# Step 7: Print Results
print("Two-Proportion Z-Test Results")
print(f"Proportion 1: {p1:.4f}")
print(f"Proportion 2: {p2:.4f}")
print(f"Z-Score: {z_score:.4f}")
print(f"P-Value: {p_value:.4f}")

# Step 8: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. The proportions are significantly different.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference in proportions.")
````

````python
# 26.  Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results.

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Generate Two Independent Normal Distributions
np.random.seed(42)

data1 = np.random.normal(loc=50, scale=10, size=30)  # Mean=50, Std=10
data2 = np.random.normal(loc=50, scale=15, size=30)  # Mean=50, Std=15

# Step 2: Compute Sample Variances
var1 = np.var(data1, ddof=1)  # Sample variance with Bessel's correction
var2 = np.var(data2, ddof=1)

# Step 3: Compute F-Statistic
if var1 > var2:  # Ensure F is always ≥ 1 by dividing larger variance by smaller one
    F_statistic = var1 / var2
    df1, df2 = len(data1) - 1, len(data2) - 1
else:
    F_statistic = var2 / var1
    df1, df2 = len(data2) - 1, len(data1) - 1

# Step 4: Compute P-Value (Two-Tailed Test)
p_value = 2 * min(stats.f.cdf(F_statistic, df1, df2), 1 - stats.f.cdf(F_statistic, df1, df2))

# Step 5: Print Results
print("F-Test for Variance Comparison")
print(f"Variance 1: {var1:.4f}")
print(f"Variance 2: {var2:.4f}")
print(f"F-Statistic: {F_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")

# Step 6: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. The variances are significantly different.")
else:
    print("\nConclusion: Fail to reject H0. No significant difference in variances.")

# Step 7: Visualization - Boxplot of Data Distributions
plt.figure(figsize=(8, 6))
sns.boxplot(data=[data1, data2], palette="Set2")
plt.xticks([0, 1], ["Dataset 1", "Dataset 2"])
plt.title("Comparison of Variances using Boxplot")
plt.ylabel("Values")
plt.grid(True)
plt.show()
````

````python
# 27. Perform a Chi-square test for goodness of fit with simulated data and analyze the results.

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Step 1: Simulate Observed Data
np.random.seed(42)
categories = ["A", "B", "C", "D"]
observed_counts = np.random.randint(20, 50, size=len(categories))  # Random counts for each category

# Step 2: Define Expected Distribution (Assumed Uniform Distribution)
total_observed = sum(observed_counts)
expected_counts = [total_observed / len(categories)] * len(categories)  # Equal probability

# Step 3: Perform Chi-Square Goodness of Fit Test
chi2_stat, p_value = stats.chisquare(f_obs=observed_counts, f_exp=expected_counts)

# Step 4: Print Results
print("Chi-Square Goodness of Fit Test")
print(f"Observed Counts: {observed_counts}")
print(f"Expected Counts: {expected_counts}")
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

# Step 5: Interpretation
alpha = 0.05  # Significance level
if p_value < alpha:
    print("\nConclusion: Reject H0. The observed distribution is significantly different from the expected distribution.")
else:
    print("\nConclusion: Fail to reject H0. The observed data fits the expected distribution.")

# Step 6: Visualization - Observed vs Expected Frequencies
plt.figure(figsize=(8, 5))
plt.bar(categories, observed_counts, alpha=0.6, label="Observed", color="blue")
plt.bar(categories, expected_counts, alpha=0.6, label="Expected", color="orange")
plt.xlabel("Categories")
plt.ylabel("Counts")
plt.title("Observed vs Expected Counts")
plt.legend()
plt.grid(True)
plt.show()
````
